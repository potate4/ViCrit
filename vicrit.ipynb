```python
# eval_vicrit_colab.py
import os, io, math, json
from PIL import Image
from tqdm import tqdm
from datasets import load_dataset
from transformers import AutoProcessor
from vllm import LLM, SamplingParams

# ---------------------------
# Config
# ---------------------------
MODEL_ID = "zyang39/ViCrit-7B"
DATASET_ID = "russwang/ViCrit-Bench"
BATCH_SIZE = 1   # reduce if VRAM is tight
OUTPUT_FILE = "vicrit_predictions.jsonl"

# ---------------------------
# Prompts
# ---------------------------
multichoice_prompt = """You are provided with an image and the description corresponding to this image.
There is one hallucination in this description. Find the hallucination phrase and answer with the hallucination phrase directly in a list.
Your output should only be a list that contains the hallucination phrase you find.

Description: {}
"""

# ---------------------------
# Helper functions
# ---------------------------
def split_list(lst, n):
    """Split a list into n roughly equal chunks."""
    chunk_size = math.ceil(len(lst)/n)
    return [lst[i:i+chunk_size] for i in range(0, len(lst), chunk_size)]

def dump_to_jsonl(obj_list, path):
    with open(path, "w") as f:
        for x in obj_list:
            f.write(json.dumps(x) + "\n")

# ---------------------------
# Main evaluation
# ---------------------------
def main():
    # Load processor and model
    print("Loading processor...")
    processor = AutoProcessor.from_pretrained(MODEL_ID)
    print("Loading model...")
    model = LLM(
        model=MODEL_ID,
        tensor_parallel_size=4,
        disable_mm_preprocessor_cache=True
    )

    # Sampling parameters
    sampling_params = SamplingParams(
        temperature=0,
        max_tokens=2048
    )

    # Load dataset
    dataset = list(load_dataset(DATASET_ID, split="train"))
    n_examples = min(5, len(dataset))
    print(f"Dataset loaded. Will run {n_examples} examples.")

    final_responses = []

    # Batch evaluation
    for batch in tqdm(split_list(dataset[:n_examples], BATCH_SIZE)):
        batch_inputs = []

        for data in batch:
            # Robust image decoding
            try:
                img = Image.open(io.BytesIO(data['image'])).convert("RGB")
                if img.height < 28 or img.width < 28:
                    print("Skipping small image")
                    continue
            except Exception as e:
                print(f"Skipping example, image decode failed: {e}")
                continue

            # Format prompt
            prompt_text = data.get("changed_caption") or data.get("caption") or data.get("task") or ""
            qs = multichoice_prompt.format(prompt_text)

            # Apply chat template
            message = [
                {"role": "system", "content": "You are a helpful assistant."},
                {"role": "user", "content": [
                    {"type": "image", "min_pixels": 224*224, "max_pixels": 1280*28*28},
                    {"type": "text", "text": qs}
                ]}
            ]
            text = processor.apply_chat_template(message, tokenize=False, add_generation_prompt=True)

            batch_inputs.append({"prompt": text, "multi_modal_data": {"image": img}})

        # Generate outputs
        if not batch_inputs:
            continue
        outputs = model.generate(batch_inputs, sampling_params=sampling_params, use_tqdm=False)

        # Collect results
        for out, data in zip(outputs, batch):
            data['response'] = out.outputs[0].text
            # Remove image to save space
            data['image'] = None
            final_responses.append(data)

    # Save results
    dump_to_jsonl(final_responses, OUTPUT_FILE)
    print(f"Done! Predictions saved to {OUTPUT_FILE}")

# ---------------------------
# Entry point
# ---------------------------
if __name__ == "__main__":
    main()
```

```python
# eval_one_fixed_colab_final.py
import os, io, time
from PIL import Image
from datasets import load_dataset
import torch

# Reduce noisy TensorFlow logs
os.environ.setdefault("TF_CPP_MIN_LOG_LEVEL", "2")

from transformers import (
    AutoTokenizer,
    AutoProcessor,
    BitsAndBytesConfig,
    logging as hf_logging
)

# Try to import the Qwen2.5 VL class
try:
    from transformers import Qwen2_5_VLForConditionalGeneration as QwenModelClass
    HAVE_QWEN_CLASS = True
except Exception:
    HAVE_QWEN_CLASS = False

hf_logging.set_verbosity_error()

MODEL_ID = "zyang39/ViCrit-7B"
DATASET_ID = "russwang/ViCrit-Bench"
USE_GPU = torch.cuda.is_available()


def build_bnb_config():
    return BitsAndBytesConfig(
        load_in_4bit=True,
        bnb_4bit_quant_type="nf4",
        bnb_4bit_use_double_quant=True,
        bnb_4bit_compute_dtype=torch.float16
    )


def load_model_and_processor():
    print("Loading tokenizer & processor (may download ~10s-100s MB)...")
    tokenizer = AutoTokenizer.from_pretrained(MODEL_ID, trust_remote_code=True)
    processor = AutoProcessor.from_pretrained(MODEL_ID, trust_remote_code=True)

    bnb = build_bnb_config()

    print("Loading model (4-bit). This can take a minute and stream checkpoint shards...")
    if HAVE_QWEN_CLASS:
        model = QwenModelClass.from_pretrained(
            MODEL_ID,
            quantization_config=bnb,
            device_map="auto",
            trust_remote_code=True,
            use_safetensors=True
        )
    else:
        from transformers import AutoModelForCausalLM
        model = AutoModelForCausalLM.from_pretrained(
            MODEL_ID,
            quantization_config=bnb,
            device_map="auto",
            trust_remote_code=True,
            use_safetensors=True
        )

    model.eval()
    print("Model loaded.\n")
    return tokenizer, processor, model


def main():
    tokenizer, processor, model = load_model_and_processor()
    dataset = load_dataset(DATASET_ID, split="train")
    n = min(5, len(dataset))
    print(f"Dataset loaded. Will run {n} examples.\n")

    device = "cuda" if USE_GPU else "cpu"

    for i in range(n):
        ex = dataset[i]

        # Make sure prompt_text is never empty
        prompt_text = ex.get("caption") or ex.get("task") or ex.get("changed_caption") or "Describe this image."

        # Robust image loading and resizing
        try:
            img = Image.open(io.BytesIO(ex["image"])).convert("RGB")
            img = img.resize((224, 224))  # resize to avoid huge pixel_values
        except Exception as e:
            print(f"Skipping example {i}, could not decode image: {e}")
            continue

        # Processor: converts text + image to model inputs
        inputs = processor(
            text=prompt_text,
            images=img,
            return_tensors="pt",
            padding="max_length",
            truncation=True
        )

        # Skip if processor produced empty tensors
        if inputs["input_ids"].numel() == 0 or inputs["pixel_values"].numel() == 0:
            print(f"Skipping example {i}, empty input_ids or pixel_values")
            continue

        # Move tensors to GPU/CPU
        inputs = {k: v.to(device) if torch.is_tensor(v) else v for k, v in inputs.items()}

        # Debug: print shapes
        print(f"Example {i} shapes -> input_ids: {inputs['input_ids'].shape}, pixel_values: {inputs['pixel_values'].shape}")

        # Generate prediction
        with torch.no_grad():
            out = model.generate(**inputs, max_new_tokens=64)

        pred = tokenizer.decode(out[0], skip_special_tokens=True)
        print(f"=== Example {i+1} ===")
        print("GT caption:", ex.get("caption"))
        print("Prediction:", pred)
        print()


if __name__ == "__main__":
    start = time.time()
    try:
        main()
    except KeyboardInterrupt:
        print("Interrupted by user.")
    except Exception as e:
        print("Error during run:", e)
        print("Check image preprocessing and processor output.")
    finally:
        print("Done. Elapsed:", time.time() - start)
```
